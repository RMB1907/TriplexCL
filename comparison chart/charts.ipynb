```python
import pandas as pd
import matplotlib.pyplot as plt

# Load your file
df = pd.read_csv("rna.csv")

# Plot
plt.figure()

plt.plot(df["Label"], df["TriplexCL"], marker='o', label="TriplexCL")
plt.plot(df["Label"], df["TriplexFPP"], marker='s', label="TriplexFPP")

# Formatting
plt.xlabel("lncRNA Transcript")
plt.ylabel("Triplex-Forming Potential")
plt.title("Comparison of TriplexCL and TriplexFPP on held out test set")

plt.xticks(rotation=45)
plt.legend()
#plt.tight_layout()

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load your file
df = pd.read_csv("dna.csv")

# Plot
plt.figure()

plt.plot(df["Label"], df["TriplexCL"], marker='o', label="TriplexCL")
plt.plot(df["Label"], df["TriplexFPP"], marker='s', label="TriplexFPP")

plt.xticks(rotation=60, fontsize=6)
# plt.xticks(df.index, df["Label"], rotation=60, fontsize=8)
# Formatting
plt.xlabel("DNA Binding Site")
plt.ylabel("Triplex-Forming Potential")
plt.title("Comparison of TriplexCL and TriplexFPP on held out test set")

plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()

plt.show()
```
